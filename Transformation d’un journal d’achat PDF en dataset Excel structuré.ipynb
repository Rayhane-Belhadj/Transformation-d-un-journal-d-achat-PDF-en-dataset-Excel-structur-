```python
import pdfplumber
import pandas as pd

pdf_file_path = 'JOURNAL ACHAT DETAILLE19085-24.pdf'
output_excel_path = 'JJJOURNAL_ACHAT_DETAILLE19085_244_Excel.xlsx'

columns = ["Num Livr", "Date", "Num BI", "Fournisseur", 
           "Article", "Qte", "Prix", "Mnt THT", "TVA", "Mnt TVA"]

data = []

with pdfplumber.open(pdf_file_path) as pdf:
    for page_num, page in enumerate(pdf.pages, start=1):
        lines = page.extract_text().split('\n')
        header_info = None
        
        for i, line in enumerate(lines):
            print(f"Page {page_num}, Line {i}: {line}")
            
            if "Num Livr" in line and "Date" in line:
                header_info = line.split()
                print(f"Extracted Header Info: {header_info}")

            elif header_info and any(char.isdigit() for char in line):
                row = line.split()
                
                try:
                    num_livr = row[0]
                    date = row[1]
                    num_bi = row[2]
                    fournisseur = " ".join(row[3:5])
                    article = " ".join(row[5:-5])
                    qte = row[-5]
                    prix = row[-4]
                    mnt_tht = row[-3]
                    tva = row[-2]
                    mnt_tva = row[-1]

                    data.append([num_livr, date, num_bi, fournisseur, article, qte, prix, mnt_tht, tva, mnt_tva])
                except IndexError:
                    print(f"Skipped malformed line: {line}")

df = pd.DataFrame(data, columns=columns)

df.to_excel(output_excel_path, index=False)

print(f"Processed data saved to {output_excel_path}")
```
